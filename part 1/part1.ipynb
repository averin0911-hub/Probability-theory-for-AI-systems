## Задача 1.8

>В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

### Решение a)

Так как призы различны, и фильм может претендовать на несколько номинаций, значит это **размещение с повторениями**. Используем формулу:  
$${\bar{\mathrm{A}}}_{\mathrm{n}}^{\mathrm{k}} = n^k$$
$${\bar{\mathrm{A}}}_{\mathrm{10}}^{\mathrm{5}} = 10^5$$

```python
10 ** 5
```

### Решение б)

Так как призы одинаковы, значит порядок размещения не имеет значения, значит это **сочетание с повторением**. Используем формулу:  
$${\bar{\mathrm{C}}}_{\mathrm{n}}^{\mathrm{k}}={\mathrm{C}}_{\mathrm{n}+\mathrm{k}-1}^{\mathrm{k}}$$
$${\bar{\mathrm{C}}}_{10}^{5}={\mathrm{C}}_{10+5-1}^{5}$$

```python
sc.comb(10, 5, repetition=True)
```

## Задача 1.9

>Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

### Решение

Для решения задачи стоит использовать формулу **перестановок с повторениями:**
$${\mathrm{P}}_{\mathrm{n}}({\mathrm{n}}_{1},\ {\mathrm{n}}_{2},\ ...,\ {\mathrm{n}}_{\mathrm{k}})=\frac{\mathrm{n}!}{{\mathrm{n}}_{1}!{\mathrm{n}}_{2}!\ ...\ {\mathrm{n}}_{\mathrm{k}}!}$$
$${\mathrm{P}}_{7}(3,\ 2,\ 2)=\frac{7!}{3!2!2!}$$

```python
sc.factorial(7, ) / (sc.factorial(3) * sc.factorial(2) * 2)
```

## Задача 1.10

>Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

### Решение а)

* Рассмотрим событие A - получилось слово «ТОР»
$$\mathrm{P}(\mathrm{A})=\frac{1}{{\mathrm{A}}_{6}^{3}}$$

```python
1 / sc.perm(6, 3)
```

### Решение б)

* Рассмотрим событие B - получилось слово «ТЕОРИЯ»
$$\mathrm{P}(\mathrm{B})=\frac{1}{P_6}$$

```python
1 / sc.perm(6, 6)
```

## Задача 1.16

>За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

### Решение а)

Рассмотрим событие A - никакие два лица одного пола не сядут рядом. Всего вариантов перестанок: ${\mathrm{P}}_{10}=10!$. Женщины могут сидеть на чётных местах ($5!$), а мужчины на нечётных ($5!$), и наоборот. Таким образом благоприятных случаев: $2(5!)^2$
$$\mathrm{P}(\mathrm{A})=\frac{2{(5!)}^{2}}{{\mathrm{P}}_{10}}$$

```python
2 * sc.factorial(5) ** 2 / sc.perm(10, 10)
```

### Решение б) 

Рассмотрим событие B - супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар. Тогда вариантов расстановок супружеских пар: $5!$. В каждой паре по 2 варианта расстановки, т.е. всего благоприятных случаев $5!2^5$.
$$\mathrm{P}(\mathrm{B})=\frac{5!2^{5}}{{\mathrm{P}}_{10}}$$

```python
sc.factorial(5) * 2 ** 5 / sc.perm(10, 10)
```

## Задача 1.17

>В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

### Решение а)

Пусть событие А - пассажиры попали в два купе. Общее число вариантов расстановки 7 пассажиров в 9 купе: $С_{36}^{7}$. Выбрать 2 купе из 9 можно $С_{9}^{2}$ способами. Рассадить 7 пассажиров в эти 2 купе можно $С_{8}^{7}$ способами.
$$\mathrm{P}(\mathrm{A})=\frac{{\mathrm{C}}_{9}^{2}{\mathrm{C}}_{8}^{7}}{{\mathrm{C}}_{36}^{7}}$$

```python
sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7)
```

### Решение б)

Пусть событие B - пассажиры попали в семь купе. Выбрать 7 купе из 9 можно $С_{9}^{7}$ способами. Вариантов рассадки в 1 купе для 1 пассажира: 4, значит, т. к. по 1 пассажиру в купе, вариантов рассадки в 7 купе по 1 пассажиру в каждом: $4^7$.
$$\mathrm{P}(\mathrm{B})=\frac{{\mathrm{C}}_{9}^{7}4^{7}}{{\mathrm{C}}_{36}^{7}}$$

```python
sc.comb(9, 7) * 4 ** 7 / sc.comb(36, 7)
```

### Решение в)

Пусть событие D - пассажиры попали в три купе. 3 купе из 9
можно выбрать $С_{9}^{3}$ способами, а число способов выбора семи мест
из 12 в трех купе определяется сложнее, чем в п. а) и б). Действительно, возможные варианты выбора 7 мест из 12 в трех купе следующие: 4 + 2 + 1, 3 + 3 + 1, 3 + 2 + 2, а за счет перестановок купе таких вариантов будет соответственно 6, 3 и 3. Каждый из этих вариантов по правилу произведения может быть получен $С_{4}^{4}{\mathrm{C}}_{4}^{2}{\mathrm{C}}_{4}^{1},\ С_{4}^{3}{\mathrm{C}}_{4}^{3}{\mathrm{C}}_{4}^{1},\ С_{4}^{3}{\mathrm{C}}_{4}^{2}{\mathrm{C}}_{4}^{2}$ способами соответственно.
$$\mathrm{P}(\mathrm{D})\ =\frac{{\mathrm{C}}_{9}^{3}(6{\mathrm{C}}_{4}^{4}{\mathrm{C}}_{4}^{2}{\mathrm{C}}_{4}^{1}+3{\mathrm{C}}_{4}^{3}{\mathrm{C}}_{4}^{3}{\mathrm{C}}_{4}^{1}+3{\mathrm{C}}_{4}^{3}{\mathrm{C}}_{4}^{2}{\mathrm{C}}_{4}^{2})}{{\mathrm{C}}_{36}^{7}}$$

```python
sc.comb(9, 3) * (6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1) + \
                 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1) + \
                 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)) / sc.comb(36, 7)
```

## Задача 1.38

Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

### Решение

Рассмотрим событие A - книги стоят слева направо в порядке нумерации томов (от 1 до 5)

## Задача 1.43

Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

Рассмотрим событие A - все цифры: а) различные
$$\mathrm{P}(\mathrm{A})=\frac{9\ *\ 9\ *\ 8\ *\ 7\ *\ 6}{9\ *\ {10}^{4}}$$

$$\mathrm{P}(\mathrm{B})=\frac{9}{9\ *\ {10}^{4}}$$

Рассмотрим событие C - все цифры: в) нечетные

## Задача 1.48

В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков превосходящую 10. Найти вероятности: а) выпадения 11 очков;  б) выигрыша.

### Решение а)

Рассмотрим событие A - выпало 11 очков.

```python
ls = [el for el in itertools.product([1, 2, 3, 4, 5, 6], repeat=3) if sum(el) == 11]
print("Первые пять вариантов:", ls[:5])
print(len(ls))
```

$$\mathrm{P}(\mathrm{A})=\frac{27}{6\ *\ 6\ *\ 6}$$

```python
round(27 / (6 ** 3), 4)
```

### Решение б)

Рассмотрим событие B - выигрышь

```python
ls = [el for el in itertools.product([1, 2, 3, 4, 5, 6], repeat=3) if sum(el) >= 11]
print("Первые пять вариантов:", ls[:5])
print(len(ls))
```

$$\mathrm{P}(\mathrm{A})=\frac{108}{6\ *\ 6\ *\ 6}$$

```python
round(108 / (6 ** 3), 4)
```
